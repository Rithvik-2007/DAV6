```python
import os
import numpy as np
import pandas as pd
from scipy.stats import kurtosis, skew

# 1. Dataset File Locations / URLs
uci_path = "uci_diabetes.csv"
pima_path = "pima_diabetes.csv"

# Fallback download URLs if files don't exist locally
uci_url = "https://archive.ics.uci.edu/ml/machine-learning-databases/00296/dataset_diabetes.zip"  # Standard Pima URL works well for demo
pima_url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/pima-indians-diabetes.data.csv"

pima_cols = [
    "Pregnancies",
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI",
    "DiabetesPedigreeFunction",
    "Age",
    "Outcome",
]

# Helper to load dataset safely
if os.path.exists(uci_path):
    uci_diabetes = pd.read_csv(uci_path)
else:
    print(f"'{uci_path}' not found locally. Loading fallback data...")
    # Loading standard Pima structure with column headers
    uci_diabetes = pd.read_csv(pima_url, names=pima_cols)

if os.path.exists(pima_path):
    pima_diabetes = pd.read_csv(pima_path)
else:
    print(f"'{pima_path}' not found locally. Loading fallback data...")
    pima_diabetes = pd.read_csv(pima_url, names=pima_cols)

# 2. Display Dataset Samples
print("UCI Diabetes Dataset Sample:")
print(uci_diabetes.head())

print("\nPima Indians Diabetes Dataset Sample:")
print(pima_diabetes.head())

# 3. Define Relevant Numerical Columns
numerical_columns = [
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI",
    "DiabetesPedigreeFunction",
    "Age",
]


# 4. Univariate Analysis Function (Fixed Indentation)
def univariate_analysis(df, columns):
    stats = {}
    for col in columns:
        if col in df.columns:
            # Drop NaN values before computing stats to prevent errors
            clean_series = df[col].dropna()
            stats[col] = {
                "Mean": np.mean(clean_series),
                "Median": np.median(clean_series),
                "Mode": (
                    clean_series.mode()[0] if not clean_series.empty else np.nan
                ),
                "Variance": np.var(clean_series, ddof=1),
                "Standard Deviation": np.std(clean_series, ddof=1),
                "Skewness": skew(clean_series),
                "Kurtosis": kurtosis(clean_series),
            }
    return pd.DataFrame(stats).T


# 5. Perform Univariate Analysis
uci_stats = univariate_analysis(uci_diabetes, numerical_columns)
pima_stats = univariate_analysis(pima_diabetes, numerical_columns)

# 6. Display Results
print("\nUCI Diabetes Dataset Statistics:")
print(uci_stats)

print("\nPima Indians Diabetes Dataset Statistics:")
print(pima_stats)
```
